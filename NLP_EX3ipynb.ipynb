```python
pattern = r'\s'
data['char_count']= data['twitts'].apply(lambda x:len(re.sub(pattern,'',x)))
```

```python
data['word_count']= data['twitts'].apply(lambda x: len(x.split()))
```

```python
data['avg_word_length'] = data['char_count']/data['word_count']
```
